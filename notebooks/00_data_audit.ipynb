# 00 — Data audit trước tiền xử lý

Notebook này đọc báo cáo đã chạy trên toàn bộ `delivery_yt.csv`. Không được chọn ngưỡng lọc trước khi xem các bảng profiling, cohort sensitivity và anomaly examples.


```python
from pathlib import Path
import json
import pandas as pd

report = Path('../reports/data')
summary = json.loads((report/'summary.json').read_text(encoding='utf-8'))
print(json.dumps(summary, ensure_ascii=False, indent=2))

for name in ['column_profile','quantiles','quality_flags','cohort_sensitivity','split_profile','courier_day','aoi_relations']:
    path = report/'tables'/f'{name}.csv'
    print(f'\n=== {name} ===')
    display(pd.read_csv(path).head(30))
```

```python
quality = pd.read_csv(report/'row_quality.csv.gz')
print('row_quality rows:', len(quality))
display(quality.sample(min(20, len(quality)), random_state=42))
display(pd.read_csv(report/'tables/anomaly_examples.csv').head(30))
```
