### 1. Đọc mẫu từ file gốc

File đầu vào là `data/delivery_yt.csv`. Mỗi dòng là một đơn giao hàng.
Trong Docker, Jupyter mở file này tại `/home/jovyan/data/delivery_yt.csv`.
Xem giờ nhận, giờ giao và GPS của ba dòng trước khi bắt đầu xử lý.

```python
from pathlib import Path
import sys

import pandas as pd

cwd = Path.cwd().resolve()
ROOT = next((p for p in (cwd, cwd.parent, Path("/home/jovyan/work"))
             if (p / "scripts" / "profile_delivery.py").is_file()), None)
if ROOT is None:
    raise FileNotFoundError("Mở notebook từ thư mục gốc dự án, thư mục notebooks hoặc Jupyter Docker.")
sys.path.insert(0, str(ROOT))

source = ROOT / "data" / "delivery_yt.csv"
if not source.is_file():
    source = ROOT.parent / "data" / "delivery_yt.csv"
preview = pd.read_csv(source, dtype=str, keep_default_na=False, nrows=3)
print(source)
print(f"{len(preview.columns)} cột")
print(preview[["order_id", "courier_id", "accept_time", "delivery_time",
               "accept_gps_time", "delivery_gps_time"]].to_string(index=False))
```
